```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
sample_values = np.array([5.2, 5.4, 5.1, 5.6, 5.3, 5.8])
target_mean = 5.5
```

```python
test_stat, p_val = stats.ttest_1samp(sample_values, target_mean)
conf_int = stats.t.interval(
    0.95,
    df=len(sample_values) - 1,
    loc=np.mean(sample_values),
    scale=stats.sem(sample_values)
)
```

```python
print("Sample Mean:", np.mean(sample_values))
print("Hypothesized Mean:", target_mean)
print("t-Statistic:", test_stat)
print("Degrees of Freedom:", len(sample_values) - 1)
print("p-Value:", p_val)
print("Confidence Interval:", conf_int)
```

```python

alpha_level = 0.05
if p_val < alpha_level:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

```python

plt.figure()
plt.bar(1, np.mean(sample_values), color="green", label="Sample Mean")
plt.errorbar(1, np.mean(sample_values), yerr=np.std(sample_values), fmt='k', linewidth=2)
plt.axhline(y=target_mean, color='r', linestyle='--', label='Hypothesized Mean')
plt.title('Sample Mean vs Hypothesized Mean')
plt.xlim(0.5, 1.5)
plt.xlabel('Sample')
plt.ylabel('Mean Value')
plt.legend()
plt.grid(True)
plt.show()
```
